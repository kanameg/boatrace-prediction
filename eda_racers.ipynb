```python
# データの読み込み
racers_df = pd.read_csv("data/racers.csv")
```

```python
# 2025年2期のデータ抽出
racers20252_df = racers_df[(racers_df["年"] == 2025) & (racers_df["期"] == 2)]
```

```python
racers20252_df.head()
```

```python
# 1着回数の分布
plt.figure(figsize=(18, 6))
sns.histplot(
    data=racers20252_df,
    x="1着回数",
    bins=range(0, racers20252_df["1着回数"].max() + 2),
    kde=False,
)
plt.title("2025年2期 1着回数の分布")
plt.xlabel("1着回数")
plt.ylabel("選手数")
plt.xticks(range(0, racers20252_df["1着回数"].max() + 1))
plt.grid(axis="y")
plt.show()
```

```python
# 2着回数の分布
plt.figure(figsize=(18, 6))
sns.histplot(
    data=racers20252_df,
    x="2着回数",
    bins=range(0, racers20252_df["2着回数"].max() + 2),
    kde=False,
)
plt.title("2025年2期 2着回数の分布")
plt.xlabel("2着回数")
plt.ylabel("選手数")
plt.xticks(range(0, racers20252_df["2着回数"].max() + 1))
plt.grid(axis="y")
plt.show()
```

```python
# ------------------------------------------------------------------------------------------------------------------------
# 出走回数がまちまちなので、出走回数に対する割合を計算
# ------------------------------------------------------------------------------------------------------------------------
```

```python
racers20252_df["1着率"] = racers20252_df["1着回数"] / racers20252_df["出走回数"]
racers20252_df["2着率"] = racers20252_df["2着回数"] / racers20252_df["出走回数"]
# 1コース進入率
racers20252_df["1コース進入率"] = (
    racers20252_df["1コース進入回数"] / racers20252_df["出走回数"]
)
# 2コース進入率
racers20252_df["2コース進入率"] = (
    racers20252_df["2コース進入回数"] / racers20252_df["出走回数"]
)
# 3コース進入率
racers20252_df["3コース進入率"] = (
    racers20252_df["3コース進入回数"] / racers20252_df["出走回数"]
)
# 4コース進入率
racers20252_df["4コース進入率"] = (
    racers20252_df["4コース進入回数"] / racers20252_df["出走回数"]
)
# 5コース進入率
racers20252_df["5コース進入率"] = (
    racers20252_df["5コース進入回数"] / racers20252_df["出走回数"]
)
# 6コース進入率
racers20252_df["6コース進入率"] = (
    racers20252_df["6コース進入回数"] / racers20252_df["出走回数"]
)
```

```python
racers20252_df.head()
```

```python
# 1着率の分布
plt.figure(figsize=(18, 6))
sns.histplot(
    data=racers20252_df,
    x="1着率",
    bins=np.arange(0, 1.05, 0.05),
    kde=False,
)
plt.title("2025年2期 1着率の分布")
plt.xlabel("1着率")
plt.ylabel("選手数")
plt.xticks(np.arange(0, 1.05, 0.05))
plt.grid(axis="y")
plt.show()
```

```python
# 優秀選手の分析
top_racers = racers20252_df[racers20252_df["1着回数"] >= 30].copy()
top_racers.sort_values(by="1着回数", ascending=False, inplace=True)
print("------------------------------------------------------------")
print(f"優秀選手数: {len(top_racers)}人")
print("------------------------------------------------------------")
# top_racers.head()
```

```python
# 通常の選手の分析
normal_racers = racers20252_df[
    (racers20252_df["1着回数"] >= 5) & (racers20252_df["1着回数"] < 10)
].copy()
normal_racers.sort_values(by="1着回数", ascending=False, inplace=True)
print("------------------------------------------------------------")
print(f"通常選手数: {len(normal_racers)}人")
print("------------------------------------------------------------")
# normal_racers.head()
```

```python
# 平均スタートタイミングの比較
plt.figure(figsize=(12, 6))
sns.kdeplot(
    data=top_racers,
    x="平均スタートタイミング",
    fill=True,
    label="優秀選手 (1着回数 >= 30)",
    color="blue",
    alpha=0.5,
)
sns.kdeplot(
    data=normal_racers,
    x="平均スタートタイミング",
    fill=True,
    label="通常選手 (5 <= 1着回数 < 10)",
    color="orange",
    alpha=0.5,
)
plt.title("平均スタートタイミングの分布比較")
plt.xlabel("平均スタートタイミング")
plt.ylabel("密度")
plt.legend()
plt.grid()
plt.show()
```

```python
# 優秀選手と通常選手の級のcountplot
plt.figure(figsize=(12, 6))
sns.countplot(
    data=pd.concat(
        [
            top_racers.assign(カテゴリ="優秀選手"),
            normal_racers.assign(カテゴリ="通常選手"),
        ]
    ),
    x="級",
    hue="カテゴリ",
)
plt.title("級別選手数の比較")
plt.xlabel("級")
plt.ylabel("選手数")
plt.grid(axis="y")
plt.show()
```
