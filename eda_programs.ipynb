```python
# データの読み込み
programs_df = pd.read_csv("data/programs.csv")
```

```python
# 年,月,日を結合して日付型に変換
programs_df.insert(
    0,
    "開催日",
    pd.to_datetime(
        programs_df["年"].astype(str)
        + "-"
        + programs_df["月"].astype(str)
        + "-"
        + programs_df["日"].astype(str)
    ),
)
programs_df.head()
```

```python
# 2025-01-01から2025-06-30までのデータを抽出
programs20251_df = programs_df[
    (programs_df["開催日"] >= "2025-01-01") & (programs_df["開催日"] <= "2025-06-30")
].copy()
print("------------------------------------------------------------")
print(f"2025年上半期のレース数: {len(programs20251_df)}レース")
print("------------------------------------------------------------")
programs20251_df.tail()
```

```python
# 級別ごとに1から6の枠番回数が何回あるかを集計
frame_counts = programs20251_df.groupby(["級別", "枠番"]).size().unstack(fill_value=0)
frame_counts.columns = [f"{col}枠" for col in frame_counts.columns]
frame_counts.plot(kind="bar", stacked=False, figsize=(10, 6))
plt.title("2025年上半期 級別ごとの枠番分布")
plt.xlabel("級別")
plt.ylabel("出走回数")
plt.legend(title="枠番", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
